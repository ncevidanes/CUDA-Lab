# CUDA-Lab — ROOT + CUDA multi-file
Pipeline: Drive → ROOT/CPU → pinned buffers → CUDA → CPU/GPU gates → Colab + Drive.


```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os, subprocess, pathlib
REPO = '/content/CUDA-Lab'
REF = os.environ.get('CUDA_LAB_REF', 'feature/root-cuda-multifile')
if pathlib.Path(REPO).exists():
    subprocess.run(['rm','-rf',REPO], check=True)
subprocess.run(['git','clone','https://github.com/ncevidanes/CUDA-Lab.git',REPO], check=True)
subprocess.run(['git','-C',REPO,'checkout',REF], check=True)
```

```python
subprocess.run(['bash', f'{REPO}/scripts/colab_root_cuda_bootstrap.sh'], check=True, env={**os.environ, 'REPO_ROOT': REPO})
```

```python
env = {**os.environ, 'REPO_ROOT': REPO}
env.setdefault('ROOT_INPUT_DIR', '/content/drive/MyDrive/CUDA-Lab/root-input')
env.setdefault('ROOT_DRIVE_OUTPUT_BASE', '/content/drive/MyDrive/CUDA-Lab/root-results')
subprocess.run(['bash', f'{REPO}/scripts/colab_root_cuda_run.sh'], check=True, env=env)
```
